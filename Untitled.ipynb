```python
#sort math from highest grade to lowest
top_math = math_grades.sort_values(by = "Grade", ascending = False)
top_math
```

```python
#marks between 80 and 90
filtered_df = grades[(grades['Grade'] >= 80) & (grades['Grade'] <= 90)]
filtered_df
```

```python
#same as above but only where grade is 88 or 85
filtered_df = grades[(grades['Grade'] == 88) | (grades['Grade'] == 85)]
filtered_df
```

```python
filtered_df = grades[grades['Grade'].isin([88,85])]
filtered_df
```

```python
#bnumber of stuents in each class
grades["Class"].value_counts()
```

```python
import matplotlib.pyplot as plt
```

```python
#average mark for each class
avg_by_class = grades.groupby("Class")["Grade"].mean()
avg_by_class
```

```python
avg_by_class.plot(kind = "bar",title = "Average grade by class")
plt.ylabel("Average Grade")
plt.xticks(rotation = 45)
```

```python
#'%1.1f%%' is a format string:

# % → start formatting
# 1.1f →
# 1 = minimum width (not very important here)
# .1 = 1 decimal place
# f = floating point number
# %% → literal % sign (you need two % to show one %)
grades['Class'].value_counts().plot( kind='pie', autopct='%1.1f%%',title='Students per Class')
#plt.ylabel("")  # removes awkward label
plt.show()
```
