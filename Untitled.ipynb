```python
from pathlib import Path
```

```python
base_dir = Path.cwd()
csv_path = base_dir/"student_grades.csv"
```

```python
import pandas as pd#Import pandas package and give it an alias of pd
```

```python
#load the contents of the csv file into a dataframe
#a dataframe is a 2 dimensional view of a csv file
grades = pd.read_csv(csv_path)
```

```python
grades.head()#shows the first 5 rows of a dataframe
```

```python
grades.info()#summary of dataframe info
```

```python
grades.shape #how many rows, columns
```

```python
grades.describe()#some statistics
```

```python
#show only the math records
#Return the records where the grades["Class"} == "Math" is True
math_grades = grades[grades["Class"] == "Math"]
math_grades
```

```python
#how many math students are there?
math_grades.shape[0]
```

```python
#sort math from highest grade to lowest
top_math = math_grades.sort_values(by = "Grade", ascending = False)
top_math
```

```python
#top math student record
top_math.iloc[0]
```

```python
#Exercise!
#Get the science grades and display them
science_grades = grades[grades["Class"] == "Science"]
science_grades
#sort the science grades from highest to lowest
top_science = science_grades.sort_values(by = "Grade", ascending = False)
top_science
#Display the row with the highest science grade
top_science.iloc[0]
```
